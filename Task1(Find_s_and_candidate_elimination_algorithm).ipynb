```python
import numpy as np
import pandas as pd

# Load data
data = pd.read_csv('loan_data_set.csv')

# All columns except last are attributes (concepts)
concepts = np.array(data.iloc[:, 0:-1])
print("\nInstances are:\n", concepts)

# Last column is target (Loan Approved)
target = np.array(data.iloc[:, -1])
print("\nTarget Values are: ", target)


def learn(concepts, target):
    # Initialize specific and general hypotheses
    specific_h = concepts[0].copy()
    print("\nInitialization of specific_h and general_h")
    print("\nSpecific Boundary: ", specific_h)
    general_h = [["?" for _ in range(len(specific_h))]
                 for _ in range(len(specific_h))]
    print("\nGeneric Boundary: ", general_h)

    for i, h in enumerate(concepts):
        print("\nInstance", i + 1, "is", h)

        if target[i].lower() == "yes":
            print("Instance is Positive")
            # Generalize specific_h
            for x in range(len(specific_h)):
                if h[x] != specific_h[x]:
                    specific_h[x] = '?'
                    general_h[x][x] = '?'

        elif target[i].lower() == "no":
            print("Instance is Negative")
            # Specialize general_h
            for x in range(len(specific_h)):
                if h[x] != specific_h[x]:
                    general_h[x][x] = specific_h[x]
                else:
                    general_h[x][x] = '?'

        print("Specific Boundary after", i + 1, "instance is", specific_h)
        print("Generic Boundary after", i + 1, "instance is", general_h)
        print()

    # Remove fully '?' rows from general_h
    to_remove = ['?'] * len(specific_h)
    general_h = [g for g in general_h if g != to_remove]

    return specific_h, general_h


s_final, g_final = learn(concepts, target)
print("\nFinal Specific Hypothesis:\n", s_final)
print("\nFinal General Hypotheses:\n", g_final)
```
